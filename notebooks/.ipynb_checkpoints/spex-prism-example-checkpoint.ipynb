# SPEX prism point-source reduction example

Last updated 23 May 2025

This notebook demonstrates the reduction of a (pre-upgraded) SPEX prism point-source data using the test data provided with the code in tests/test_data/raw/spex-prism

## Steps:
* setup instrument parameters
* reduce calibration data
* extract science point source
* extract telluric standard point source
* combine individual science spectra into one spectrum
* combine individual telluric spectra into one spectrum
* telluric calibration science spectrum


```python
# import package and check
import pyspextool as ps
import os, sys
import numpy as np
import matplotlib.pyplot as plt

help(ps)
```

# Setup

While users should be able to run the Jupyter notebook tutorials without changing the output paths, users are encouraged to move the data folder and rename the output paths. Users can define the output paths, including the `cal`, `proc`, and `qa` in the next cell to where users would like to save their data.

```python
# ============================================================================
# CONFIGURATION - Define paths, files, and reduction parameters
# ============================================================================

# Instrument configuration
instrument = 'spex'        # Pre-upgraded SpeX instrument
inst_mode  = 'prism'       # Prism mode (short wavelength, low resolution)

# Determine data paths
# file_base: Root directory of the pyspextool installation
file_base = ps.__file__.split('/src/')[0]
topbase = os.getenv("HOME")  # User's home directory

# Data directories
# raw: Location of raw FITS files from the telescope
raw = file_base + '/tests/test_data/raw/' + instrument + '-' + inst_mode + '/data/'

# Output directories (users can customize these paths)
cal  = topbase + '/test_pyspextool/test_data/processed/' + instrument + '-' + inst_mode + '/cals/' # Calibration files (flats, wavelength solutions)
proc = topbase + '/test_pyspextool/test_data/processed/' + instrument + '-' + inst_mode + '/proc/' # Processed spectra
qa   = topbase + '/test_pyspextool/test_data/processed/' + instrument + '-' + inst_mode + '/qa/'   # Quality assurance plots

# File naming prefixes (instrument-specific)
src_prefix  = 'spc'        # Science and standard star files
flat_prefix = 'flat'       # Flat field files
arc_prefix  = 'arc'        # Arc lamp (wavelength calibration) files

# ============================================================================
# INPUT FILES - Define which files to use for each step
# ============================================================================

# Target and standard star names
obj_name = 'J1104+1959'    # Science target name
std_name = 'HD101060'      # Telluric standard star name

# File ranges (format: 'start-end' or comma-separated list)
files      = '1-8'         # Science target observations
std_files  = '9-14'        # Telluric standard observations
flat_files = '15-19'       # Flat field frames
wave_files = '20'          # Arc lamp frames for wavelength calibration

# ============================================================================
# EXTRACTION PARAMETERS - Configure spectral extraction settings
# ============================================================================

# Observation mode
mode = 'A-B'               # A-B nodding (subtract beam positions)
naps = 1 if mode == 'A' else 2  # Number of apertures (1 for A, 2 for A-B)
aperture_signs = None      # Signs of apertures (None for auto-detection)

# Aperture finding and extraction
aperture_find  = ['auto',2] # Auto-find apertures with 2-sigma threshold
aperture       = 1.5        # Aperture radius in arcseconds
include_orders = None       # Which orders to extract (None = all)

# Correction options
correct_bias      = True   # Apply bias correction
correct_linearity = True   # Apply linearity correction
flat_field        = True   # Apply flat field correction
fix_badpixels     = True   # Interpolate over bad pixels

# Background subtraction
bg_annulus = [2,2.5]       # Inner and outer radius (arcsec) for background annulus
psf_radius = None          # PSF radius for optimal extraction (None = aperture photometry)

# Output files for calibrations
flatfield_file = 'flat%s.fits' % flat_files
wavecal_file   = 'wavecal%s.fits' % wave_files

# Telluric correction parameters file names
correction_type = 'A0 V'
outfile         = '{}_telluric_corr'.format(obj_name)

# ============================================================================
# QA AND OUTPUT SETTINGS - Control quality assurance plots and verbosity
# ============================================================================

verbose             = True    # Print progress messages to console
qa_show             = True    # Display QA plots (set False for batch processing)
qa_write            = True    # Save QA plots to disk
qa_extension        = '.pdf'  # File format for QA plots (.pdf, .png, etc.)
qa_showblock        = True    # Block execution while QA plots are shown
write_model_spectra = False   # Write intermediate model spectra (for debugging)

# Create output directories if they don't exist
for directory in [cal, proc, qa]:
    if not os.path.exists(directory):
        os.makedirs(directory)
        print(f'Created directory: {directory}')
```

```python
# ============================================================================
# GENERATE OUTPUT PREFIX - Create a descriptive prefix for output files
# ============================================================================
# The output prefix encodes which corrections have been applied:
# - 'A-B' or 'A': observing mode
# - '_bc': bias corrected
# - '_lc': linearity corrected  
# - '_flat': flat fielded
# - '_wc': wavelength calibrated
# - '_fxbdpx': fixed bad pixels
# Example: 'A-B_bc_lc_flat_wc_fxbdpx-' means all corrections applied

output_prefix = mode
if correct_bias is True:  output_prefix += '_bc'
if correct_linearity is True:  output_prefix += '_lc'
if flat_field is True:  output_prefix += '_flat'
if flat_field is not None:  output_prefix += '_wc'
if fix_badpixels is True:  output_prefix += '_fxbdpx'

output_prefix += '-'
print(f'Output files will be prefixed with: {output_prefix}')
```

```python
# ============================================================================
# INITIALIZE PYSPEXTOOL - Set up the reduction environment
# ============================================================================
# This function configures the global state for the reduction pipeline,
# including file paths and quality assurance settings.

ps.pyspextool_setup(instrument,          # Instrument name ('spex' or 'uspex')
                    raw_path=raw,        # Path to raw data
                    qa_path=qa,          # Path for QA outputs
                    cal_path=cal,        # Path for calibration files
                    proc_path=proc,      # Path for processed spectra
                    verbose=verbose,     # Print status messages
                    qa_write=qa_write,   # Save QA plots
                    qa_show=qa_show,     # Display QA plots
                    qa_showblock=qa_showblock)  # Block while showing plots
```

## Calibrations

```python
# ============================================================================
# CREATE CALIBRATIONS - Build flat field and wavelength solution
# ============================================================================

# STEP 1: Create master flat field
# Combines multiple flat field exposures to create a normalized flat
# that corrects for pixel-to-pixel sensitivity variations
ps.extract.make_flat([flat_prefix, flat_files],           # Input flat files
                     flatfield_file.rstrip('.fits'))       # Output file name

# STEP 2: Create wavelength calibration solution  
# Uses arc lamp emission lines to map pixel positions to wavelengths
ps.extract.make_wavecal([arc_prefix, wave_files],         # Input arc files
                        flatfield_file,                   # Flat field for tracing orders
                        wavecal_file.rstrip('.fits'))     # Output wavelength solution
```

## Extract telluric standard files

```python
# Extract standard 

ps.extract.extract(mode,
                   [src_prefix,std_files],
                   flatfield_file,
                   wavecal_file,
                   aperture_find,
                   aperture,
                   flat_field=flat_field,
                   linearity_correction=correct_linearity,
                   output_prefix=output_prefix,
                   write_rectified_orders=False,
                   aperture_signs=aperture_signs,
                   include_orders=include_orders,
                   bg_annulus=bg_annulus,
                   fix_badpixels=fix_badpixels,
                   psf_radius=psf_radius,
                   detector_info={'correct_bias':correct_bias})
```

## Extract science files

```python
# Extract science target

ps.extract.extract(mode,
                   [src_prefix,files],
                   flatfield_file,
                   wavecal_file,
                   aperture_find,
                   aperture,
                   flat_field=flat_field,
                   linearity_correction=correct_linearity,
                   output_prefix=output_prefix,
                   write_rectified_orders=False,
                   aperture_signs=aperture_signs,
                   include_orders=include_orders,
                   bg_annulus=bg_annulus,
                   fix_badpixels=fix_badpixels,
                   psf_radius=psf_radius,
                   detector_info={'correct_bias':correct_bias})
```

## Combine standard and science spectra

```python
# Combine the standard
ps.combine.combine(['A-B_bc_lc_flat_wc_fxbdpx-',std_files],std_name)
```

```python
# Combine the science target
ps.combine.combine(['A-B_bc_lc_flat_wc_fxbdpx-',files],obj_name)
```

# Perform telluric correction with science target and standard

```python
# Perform telluric correction with science target and standard

ps.telluric.telluric(obj_name+'.fits',
                     std_name+'.fits',
                     std_name,
                     'calspec'+files,
                     outfile, 
                     correction_type=correction_type,
                     write_model_spectra=write_model_spectra, 
                     qa_write=True,
                     verbose=verbose)
```

# Checking your outputs

You should find several output files in your cals and proc folders that allow you to visualize the calibrations and extractions. In addition, you can read in the spectrum and visualize it with plot_spectra.py


```python
# plot with default parameters
from pyspextool.plot.plot_spectra import plot_spectra

file = '{}.fits'.format(outfile)
ax = plot_spectra(proc+file,colors='k', ytype="flux and uncertainty", order_numbers=False)
```
